# Inverted calibration pattern generation

Original notebook: Andrés G. Marrugo, 2021. Updated to import reusable local functions.

White circles are drawn directly on a full-page black background. No SVG editing or downloaded OpenCV scripts are needed. Run this notebook from the repository root (or install the package first).


```python
from pathlib import Path
from IPython.display import SVG, display
from calibration_patterns import Pattern, PRESETS, generate_pattern, object_points

output_dir = Path("patterns")
```

## Original A3 target, inverted

8 columns × 21 rows. The OpenCV-style spacing parameter is 16 mm: horizontal pitch is 32 mm and alternating rows shift by 16 mm. Radius is spacing / 2.5 = 6.4 mm (diameter 12.8 mm). Adjacent staggered rows have a diagonal separation of 16√2 mm. SVG shapes have no stroke.


```python
pattern = Pattern(columns=8, rows=21, spacing_mm=16, radius_ratio=2.5,
                  page_width_mm=297, page_height_mm=420, kind="acircles", inverted=True)
svg_path = generate_pattern(output_dir / "acircles_8x21_16mm_a3_inverted.svg", pattern)
display(SVG(filename=str(svg_path)))
```

## Generate the layouts from the original notebook

These are editable examples, not independently validated calibration-board choices. A2 and A1 use their actual page dimensions. Each SVG also gets a JSON geometry description and CSV of planar object points in millimetres.


```python
generated = {name: generate_pattern(output_dir / f"{name}.svg", spec)
             for name, spec in PRESETS.items()}
list(generated.values())
```

## Coordinates for calibration

Points are row-major, with the first point at (0, 0, 0). Confirm that your detector's ordering and orientation match before calibration. For circle grids, pattern size is (columns, rows). For checkerboards these parameters count squares, so detected inner-corner size is (columns−1, rows−1).

Print at 100% / actual size and measure the printed spacing. Printer margins may clip the full-page background. These are planar calibration targets; structured-light projection sequences and camera–projector correspondence decoding are separate steps.


```python
points_mm = object_points(pattern)
print(f"{len(points_mm)} calibration points")
points_mm[:8]
```
